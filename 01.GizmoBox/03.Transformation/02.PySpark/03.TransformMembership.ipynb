#### QUERY AND VALIDATE MEMBERSHIPS FROM `GIZMO.BRONZE.MEMBERSHIPS_DELTA`

```python
memberships_df = spark.table('''GIZMO.BRONZE.MEMBERSHIPS_DELTA''')
display(memberships_df)
```

#### EXTRACT CUSTOMER_ID FROM `PATH` COLUMN

```python
from pyspark.sql import functions as F

# Assuming your source dataframe is named 'df'
memberships_extract_customer_id_df = memberships_df.select(
    F.regexp_extract(F.col("path"), r".*/([0-9]+)\.png$", 1).alias("customer_id"),
    F.col("content").alias("membership_card")
)
display(memberships_extract_customer_id_df)
```

#### CREATE `GIZMO.SILVER.MEMBERSHIPS`
1. CATALOG NAME: GIZMO
2. SCHEMA NAME: SILVER
3. TABLE NAME: MEMBERSHIPS_DELTA

```python
memberships_extract_customer_id_df.writeTo('gizmo.silver.memberships_delta').createOrReplace()
```

#### VALIDATE `GIZMO.SILVER.MEMBERSHIPS_DELTA`

```python
%sql
SELECT * FROM GIZMO.SILVER.MEMBERSHIPS_DELTA;
```
