```sql
SELECT 
order_id,
customer_id,
payment_method,
order_status,
item.item_id AS item_id,
item.name AS item_name,
item.category AS category,
item.price AS item_price,
item.quantity AS item_quantity
FROM
(
SELECT 
order_id,
customer_id,
payment_method,
order_status,
explode(items) AS item,
order_timestamp,
load_timestamp
 FROM CIRCUIT.SILVER.ORDERS_ENRICHED);
```

```sql
SELECT
oi.customer_id,
cust.first_name,
cust.last_name,
cust.email,
cust.telephone,
addr.address_line_1,
addr.city,
addr.state,
addr.postcode,
sum(oi.item_price * oi.item_quantity) AS total_order_amount,
count(distinct oi.order_id) AS total_orders,
sum(oi.item_quantity) AS total_items
FROM
CIRCUIT.SILVER.ORDERS_ITEMS AS oi INNER JOIN CIRCUIT.SILVER.CUSTOMERS_SCD1 AS cust
ON oi.customer_id = cust.customer_id INNER JOIN CIRCUIT.SILVER.ADDRESSES_SCD2 AS addr
ON addr.customer_id = cust.customer_id
AND addr.`__END_AT` IS NULL
GROUP BY oi.customer_id,
cust.first_name,
cust.last_name,
cust.email,
cust.telephone,
addr.address_line_1,
addr.city,
addr.state,
addr.postcode;
```
