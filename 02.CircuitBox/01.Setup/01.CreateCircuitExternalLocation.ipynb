### SET-UP PROJECT ENVIRONMENT FOR `GIZMO` DATA LAKEHOUSE

#### CREATE `STORAGE LOCATION` FOR `CIRCUIT` CONTAINER
1. CONTAINER NAME: circuit
2. ACCOUNT NAME: gizmoboxdevadls

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS circuit_external_location
URL 'abfss://circuit@gizmoboxdevadls.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `gizmobox-dev-storage-credential`)
COMMENT 'CREATING EXTERNAL LOCATION FOR CIRCUIT CONTAINER'
```

#### CREATE `CATALOG` FOR `CIRCUIT` CONTAINER
1. CATALOG NAME: CIRCUIT
2. CONTAINER NAME: gizmoboxdevadls

```sql
CREATE CATALOG IF NOT EXISTS circuit
MANAGED LOCATION 'abfss://circuit@gizmoboxdevadls.dfs.core.windows.net/'
COMMENT 'CREATING CATALOG FOR CIRCUIT';
```

#### DISPLAY THE AVAILABLE CATALOGS

```sql
SHOW CATALOGS;
```

#### CREATE SCHEMA 
1. LANDING
2. LAKEHOUSE

#### CREATE `LANDING` SCHEMA

```sql
CREATE SCHEMA IF NOT EXISTS landing
COMMENT 'CREATE LANDING SCHEMA'
MANAGED LOCATION 'abfss://circuit@gizmoboxdevadls.dfs.core.windows.net/landing';
```

```sql
CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'CREATE BRONZE SCHEMA'
MANAGED LOCATION 'abfss://circuit@gizmoboxdevadls.dfs.core.windows.net/bronze';
```

#### CREATE `LAKEHOUSE` SCHEMA

```sql
CREATE SCHEMA IF NOT EXISTS lakehouse
COMMENT 'CREATE LAKEHOUSE SCHEMA'
MANAGED LOCATION 'abfss://circuit@gizmoboxdevadls.dfs.core.windows.net/lakehouse';
```

```sql
SHOW SCHEMAS;
```

#### CREATE VOLUME

```sql
USE CATALOG circuit;
USE SCHEMA landing;
CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
LOCATION 'abfss://circuit@gizmoboxdevadls.dfs.core.windows.net/landing/operational_data'
COMMENT 'CREATE VOLUME opearation_data';
```
