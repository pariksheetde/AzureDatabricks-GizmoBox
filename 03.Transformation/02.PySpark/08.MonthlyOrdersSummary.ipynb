#### MONTHLY ORDER SUMMARY
1. TOTAL ORDERS
2. TOTAL ITEMS BOUGHT
3. TOTAL AMOUNT SPENT

#### WRITE TRANSFORMED MONTHLY ORDER SUMMARY DATA TO GOLD SCHEMA
1. CATALOG NAME: GIZMO
2. SCHEMA NAME: GOLD
3. TABLE NAME: MONTHLY_ORDERS_SUMMARY_DELTA

```python
orders_delta_df = spark.table('''GIZMO.SILVER.ORDERS_DELTA''')
display(orders_delta_df.limit(1))
```

```python
from pyspark.sql.functions import date_format, countDistinct, sum, expr

monthly_orders_summary_df = orders_delta_df.groupBy(
    date_format('transaction_timestamp', 'yyyy-MM').alias('order_month'),
    'customer_id'
).agg(
    countDistinct('order_id').alias('total_orders'),
    sum('quantity').alias('total_item_bought'),
    expr("sum(price * quantity)").alias('total_amount')
)

display(monthly_orders_summary_df)
```

```python
monthly_orders_summary_df.writeTo('GIZMO.GOLD.MONTHLY_ORDERS_SUMMARY_DELTA').createOrReplace()
```
